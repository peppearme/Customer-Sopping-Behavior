```python
#Let's fill the null values ​​with the median of review rating but first let's divide by category, and let's calculate the median by category
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.duplicated().any()
```

```python
#To make it easier to read and remember, we transform the column headers and eliminate all spaces.
df.columns=df.columns.str.lower().str.replace(' ','_')
df.columns
```

```python
df.rename(columns={'purchase_amount_(usd)':'purchase_amount'},inplace=True)
```

```python
labels=['Young','Adult','MidAge','Senior']
df['age_category']=pd.qcut(df['age'],q=4,labels=labels)
df[['age','age_category']].head(20)
```

```python
#Now we will create a column that indicates the exact days of purchase frequency
df['frequency_of_purchases'].unique()

```

```python
# df['day_frequancy_of_purchase']=df['frequency_of_purchases'].map(
# {'Fortnightly':14,
# 'Weekly':7,
# 'Annually':365,
# 'Quarterly':90,
# 'Bi-Weekly':14,
# 'Monthly':30,
# 'Every 3 Months':90})    OPPURE

frequency_mapping={'Fortnightly':14,
'Weekly':7,
'Annually':365,
'Quarterly':90,
'Bi-Weekly':14,
'Monthly':30,
'Every 3 Months':90}
df['day_frequancy_of_purchase']=df['frequency_of_purchases'].map(frequency_mapping)
```
